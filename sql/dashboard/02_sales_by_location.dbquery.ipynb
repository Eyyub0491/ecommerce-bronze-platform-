```sql
SELECT
    COALESCE(s.state_name, 'Unknown') AS state_name,
    COALESCE(s.region, 'Unknown') AS region,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(o.quantity * o.price),2) AS total_revenue

FROM dbr_dev.brazilian_ecommerce_bronze.brz_orders o

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_customers c
    ON o.customer_id = c.customer_id

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.dim_brazil_states s
    ON c.customer_state = s.state_code

GROUP BY
    COALESCE(s.state_name, 'Unknown'),
    COALESCE(s.region, 'Unknown')

ORDER BY total_revenue DESC;
```
