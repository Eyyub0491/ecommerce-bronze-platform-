```sql


-- 1. Missing customer location information

SELECT
    'Missing Customer State' AS issue_type,
    COUNT(*) AS affected_records,
    ROUND(SUM(o.quantity * o.price),2) AS impacted_revenue

FROM dbr_dev.brazilian_ecommerce_bronze.brz_orders o

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_customers c
    ON o.customer_id = c.customer_id

WHERE c.customer_state IS NULL;


-- 2. Missing product category mapping

SELECT
    'Missing Product Category' AS issue_type,
    COUNT(*) AS affected_records,
    ROUND(SUM(o.quantity * o.price),2) AS impacted_revenue

FROM dbr_dev.brazilian_ecommerce_bronze.brz_orders o

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_products p
    ON o.product_id = p.product_id

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_products_categories pc
    ON p.product_category_name = pc.product_category_name

WHERE pc.product_category_name_english IS NULL;


-- 3. Data ingestion monitoring by source

SELECT
    source,
    COUNT(*) AS total_records,
    MIN(ingestion_timestamp) AS first_ingestion,
    MAX(ingestion_timestamp) AS last_ingestion

FROM dbr_dev.brazilian_ecommerce_bronze.brz_orders

GROUP BY source;
```
