```sql
SELECT
    pc.product_category_name_english AS category_name,
    SUM(o.quantity) AS units_sold,
    ROUND(SUM(o.quantity * o.price),2) AS total_revenue

FROM dbr_dev.brazilian_ecommerce_bronze.brz_orders o

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_products p
    ON o.product_id = p.product_id

LEFT JOIN dbr_dev.brazilian_ecommerce_bronze.brz_products_categories pc
    ON p.product_category_name = pc.product_category_name

WHERE pc.product_category_name_english IS NOT NULL

GROUP BY
    pc.product_category_name_english

ORDER BY total_revenue DESC;
```
